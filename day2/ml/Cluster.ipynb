```python
    tarData = df[["Annual Income (k$)", "Spending Score (1-100)"]]
```

```python
model = KMeans(n_clusters=3)
```

```python
model
```

```python
model.fit(tarData)
```

```python
model.labels_
```

```python
model.cluster_centers_
```

```python
model.inertia_
```

```python
labels= model.labels_
sns.scatterplot(data=df,x=tarData['Annual Income (k$)'],y=tarData['Spending Score (1-100)'],hue=labels)
```

```python
n_list = []
for i in range (1,16):
  models = KMeans(n_clusters=i).fit(tarData)
  n_list.append(models.inertia_)
```

```python
n_list
```

```python
x_axis = []
for i in range(1,16):
  x_axis.append(i)

x_axis
```

```python
plt.plot(x_axis, n_list)
```

```python
finalModel = KMeans(n_clusters=5)
```

```python
finalModel.fit(tarData)
```

```python
finalModel.labels_
```

```python
cent=finalModel.cluster_centers_
```

```python
finalModel.inertia_
```

```python
X1 = df["Annual Income (k$)"]
Y1 = df["Spending Score (1-100)"]
```

```python
labels= finalModel.labels_
sns.scatterplot(data=df,x=X1,y=Y1,hue=labels)
```

```python

sns.scatterplot(data=df,x=X1,y=Y1,hue=labels)
sns.scatterplot(x=cent[:,0],y=cent[:,1], s=90, color='red', markers=True, marker='x', label='Centroids')
```
