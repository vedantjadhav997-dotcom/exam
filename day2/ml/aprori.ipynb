```python
import numpy as np
import pandas as pd
```

```python
df=pd.read_csv("Market_Basket.csv",header=None)
df
```

```python
list1=df.values.tolist()
```

```python
list1
```

```python
notNullList = []  

for element in list1:
    newList = []
    for item in element:
        if not pd.isna(item):
            newList.append(item)
    notNullList.append(newList)

notNullList
```

```python
import seaborn as sns
```

```python

import matplotlib.pyplot as plt
notNullSeries = pd.Series(notNullList)

# Get top 20 most frequent items
top20 = notNullSeries.value_counts().head(20).index

# Plot
sns.countplot(x=notNullSeries, order=top20,palette="husl")
plt.title(" ")
plt.show()
```

```python
from apyori import apriori
```

```python
rules1 = apriori(notNullList,  min_support=0.005, min_confidence = 0.02,min_length=3,max_length=5,min_lift=2)
rules1
```

```python



res1=list(rules1)
res1
```

```python
new_df1 = pd.DataFrame()
itemsList = []
supportList = []
confidenceList = []
liftList=[]

for element in res1: 
    itemsList.append(element[0])
    supportList.append(element[1])
    confidenceList.append(element[2][0][2])
    liftList.append(element[2][0][3])


new_df1["Rules"] = itemsList
new_df1["Support"] = supportList
new_df1["Confidence"] = confidenceList
new_df1["Lift"] = liftList
```
